```python
import papermill as pm
import shutil
import time
import os

# papermill is a dev dependency of this project - `uv sync` installs it.
# If not using uv, install it manually by uncommenting this line:
# !pip install papermill
```

```python
# List of notebooks to run, in order. All results will be in individual notebooks and files and not printed here.
notebooks_to_run = [
    'Illustration/01. UnitaryHousehold.ipynb',
    'Application/01. Calibration.ipynb',
    'Application/02. Comparisons.ipynb',
    'Application/03. Timings.ipynb',
    'Application/04. Counterfactuals.ipynb',
    'Appendix/Illustrate bargaining algo.ipynb',
]
notebooks_to_run = tuple(notebooks_to_run)
```

```python
# setup temporary directory for running notebooks if something fails. Original notebooks only modified if run successfully
if os.path.isdir('temp_notebooks'): 
    try:
        shutil.rmtree('temp_notebooks')
        os.mkdir('temp_notebooks')
    except:
        print(f'failed to remove "temp_notebooks"')
else:
    os.mkdir('temp_notebooks')

# define function to copy and run notebooks
def run_notebook(filename):

    try:
        temp_filename = os.path.join('temp_notebooks', filename)
        os.makedirs(os.path.dirname(temp_filename), exist_ok=True)
        # run with cwd set to the notebook's own folder, so its local imports (e.g. figure_manager, DynamicTimeAllocationModel) resolve correctly
        notebook_dir = os.path.dirname(os.path.abspath(filename))
        pm.execute_notebook(filename,temp_filename,cwd=notebook_dir,progress_bar=False)
        shutil.copy(temp_filename,f'{filename}')
        time.sleep(1)
        os.remove(temp_filename)
        
    except Exception as es:
        print(f'failed, {es}',end='')
        
```

```python
# run all notebooks
t0 = time.time()
for notebook in notebooks_to_run:
    print(f'running {notebook}')
    %time run_notebook(notebook)

# print computation time
time_taken = time.time() - t0
weeks, rest = divmod(time_taken,60*60*24*7)
days, rest = divmod(rest,60*60*24)
hours, rest = divmod(rest,60*60)
minutes, seconds = divmod(rest, 60)
print(f'Execution time: {weeks:0.0f} weeks, {days:0.0f} days, {hours:0.0f} hours, {minutes:0.0f} minutes')
```
